# Chapter 1: Machine learning landscape

## 1.1 Types of ML algorithms:

1. Supervised - Labelled training data used to predict new values
2. Unsupervised - clustering algorithms, visualization,dimensionality reduction and anomaly detection.
3. Semi-supervised - combination of labelled and unlabelled data
4. Self supervised - generating labelled dataset from unlabelled data. e.g. mask part of an image and use that for training to recover original image.
5. Transfer learning - using pre-trained models on new tasks with limited data.
6. Reinforcement learning - learning by interacting with an environment and receiving feedback in the form of rewards or penalties.

## 1.2 Challenges with ML:

1. Quantity of data needs to be large.
2. Non-representative data can lead to biased models.
3. Data quality issues such as missing values, noise, and outliers can affect model performance.
4. Irrelevant features can lead to overfitting and poor generalization.
5. Overfitting data
    a. too complex, just picking out noise
    b. doesn't generalize well to new data
6. Underfitting data means it is too simple and cannot capture the underlying patterns in the data.

## 1.3 Hyperparameter tuning:

Its very important to keep training data independent of validation data. Always start by splitting your data into training, validation, and test sets. You can test the generalisation of a model by using the test set after training and validation. When comparing multiple models you often have training, development and test sets. The development set is used to tune hyperparameters and select the best model. The test set is used to evaluate the final model's performance.

How well the model fits the data is defined using cost functions.

- RMSE (Root Mean Squared Error) measures the average magnitude of the errors between predicted and actual values. $RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$= ||y - \hat{y}||_2
- MAE (Mean Absolute Error) measures the average absolute difference between predicted and actual values. $MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$= ||y - \hat{y}||_1

$l_k$ norms are used to measure the distance between predicted and actual values. The most common norms are:

- $l_0$ norm (Hamming distance): $||y - \hat{y}||_0 = \sum_{i=1}^{n} \mathbf{1}(y_i \neq \hat{y}_i)$ = number of non-zero elements in the difference vector.
- $l_1$ norm (Manhattan distance): $||y - \hat{y}||_1 = \sum_{i=1}^{n} |y_i - \hat{y}_i|$
- $l_2$ norm (Euclidean distance): $||y - \hat{y}||_2 = \sqrt{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$
- $l_\infty$ norm (Chebyshev distance): $||y - \hat{y}||_\infty = \max_{i=1}^{n} |y_i - \hat{y}_i|$
- $l_k$ norm - ||v||_k = $(\sum_{i=1}^{n} |v_i|^k)^{1/k}$

large k focuses on large values and neglects small values. Small k focuses on small values and neglects large values. Hence RMSE is more sensitive to outliers than MAE.
